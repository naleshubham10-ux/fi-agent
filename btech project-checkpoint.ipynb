```python
# ============================================================
# MODULE 1: OBJECT DETECTION (YOLO + PyTorch)
# ============================================================

from ultralytics import YOLO

class ObjectDetectionModule:
    """
    Uses YOLOv8 (PyTorch backend) to detect objects in FI session images.
    Detects: persons, products, shelves, furniture, signboards, etc.
    Part of the AI Analysis Engine (Layer 3 of system architecture).
    """

    # Objects suggesting a SHOP environment
    SHOP_INDICATORS = {
        'bottle', 'cup', 'bowl', 'book', 'laptop',
        'backpack', 'handbag', 'suitcase', 'tv'
    }

    # Objects suggesting a RESIDENCE
    RESIDENCE_INDICATORS = {
        'bed', 'couch', 'chair', 'dining table',
        'refrigerator', 'clock', 'potted plant', 'vase'
    }

    def __init__(self, model_size: str = "yolov8n.pt"):
        """
        Args:
            model_size: yolov8n (nano, fastest) → yolov8x (extra-large, most accurate)
                        Use yolov8m or yolov8l for production deployment.
        """
        logger.info(f"Loading YOLO model: {model_size}")
        self.model = YOLO(model_size)   # Auto-downloads on first run
        self.conf  = cfg.YOLO_CONFIDENCE
        logger.info("YOLO ready.")

    def detect(self, image_path: str) -> ObjectDetectionResult:
        """
        Run YOLOv8 inference on a single captured image.

        Args:
            image_path:r"C:\data\picture\WhatsApp Image 2026-04-12 at 6.26.05 PM.jpeg"
        Returns:
            ObjectDetectionResult
        """
        start = time.time()
        result = ObjectDetectionResult()

        try:
            predictions = self.model(image_path, conf=self.conf, verbose=False)

            for pred in predictions:
                for box in pred.boxes:
                    label = pred.names[int(box.cls)]
                    conf  = float(box.conf)
                    bbox  = box.xyxy[0].tolist()  # [x1, y1, x2, y2]

                    result.raw_detections.append({
                        "label":      label,
                        "confidence": round(conf, 3),
                        "bbox":       [round(v, 1) for v in bbox]
                    })
                    result.detected_objects.append(label)
                    result.object_counts[label] = result.object_counts.get(label, 0) + 1

            result.has_person = 'person' in result.object_counts

        except Exception as e:
            logger.error(f"YOLO detection error: {e}")

        result.inference_time_s = round(time.time() - start, 3)
        logger.info(f"YOLO: {len(result.detected_objects)} objects in {result.inference_time_s}s")
        return result

    def get_fi_signals(self, result: ObjectDetectionResult,
                        declared: PremisesType) -> Dict[str, Any]:
        """
        Evaluate YOLO output against declared premises type.
        Feeds into Fraud Signal Layer.
        """
        detected_set = set(result.detected_objects)
        shop_score   = len(detected_set & self.SHOP_INDICATORS)
        home_score   = len(detected_set & self.RESIDENCE_INDICATORS)

        objects_flag  = False
        flag_reason   = None

        if declared == PremisesType.SHOP and shop_score == 0 and len(result.detected_objects) < 3:
            objects_flag = True
            flag_reason  = "Declared SHOP but very few shop-relevant objects detected by YOLO"

        elif declared == PremisesType.RESIDENCE and home_score == 0 and len(result.detected_objects) < 2:
            objects_flag = True
            flag_reason  = "Declared RESIDENCE but no residential objects detected by YOLO"

        return {
            "person_present":   result.has_person,
            "total_objects":    len(result.detected_objects),
            "unique_classes":   len(result.object_counts),
            "shop_score":       shop_score,
            "home_score":       home_score,
            "objects_flag":     objects_flag,
            "flag_reason":      flag_reason
        }

    def visualise(self, image_path: str, result: ObjectDetectionResult):
        """Draw bounding boxes on image and display."""
        img = cv2.imread(image_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        fig, ax = plt.subplots(figsize=(10, 6))
        ax.imshow(img)
        colours = plt.cm.Set3(np.linspace(0, 1, max(len(result.raw_detections), 1)))
        for i, det in enumerate(result.raw_detections):
            x1, y1, x2, y2 = det['bbox']
            rect = patches.Rectangle(
                (x1, y1), x2-x1, y2-y1,
                linewidth=2, edgecolor=colours[i % len(colours)], facecolor='none'
            )
            ax.add_patch(rect)
            ax.text(x1, y1 - 4, f"{det['label']} {det['confidence']:.2f}",
                    color='white', fontsize=8,
                    bbox=dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.7))
        ax.set_title(f"YOLO Object Detection — {len(result.raw_detections)} objects "
                     f"[{result.inference_time_s}s]", fontweight='bold')
        ax.axis('off'); plt.tight_layout(); plt.show()


print("✅ ObjectDetectionModule (YOLO) defined.")
print("\nUsage:")
print("  detector = ObjectDetectionModule('yolov8n.pt')")
print("  result   = detector.detect('shop_image.jpg')")
print("  signals  = detector.get_fi_signals(result, PremisesType.SHOP)")
```

```python
# ============================================================
# MODULE 2: OCR — EasyOCR (primary) + Tesseract (fallback)
# ============================================================

import easyocr
import pytesseract

class OCRModule:
    """
    Extracts visible text from FI session images.
    Primary: EasyOCR (supports English + Hindi scripts — Indian signboards).
    Fallback: Tesseract (AWS-deployed Tesseract as per blackbook).
    Includes fuzzy business name matching against declared name.
    """

    def __init__(self, languages: List[str] = None, gpu: bool = False):
        """
        Args:
            languages: EasyOCR language codes. Default: ['en', 'hi'] (English + Hindi)
            gpu:       Use GPU if available
        """
        self.languages = languages or ['en', 'hi']
        logger.info(f"Initialising EasyOCR — languages: {self.languages}")
        self.reader = easyocr.Reader(self.languages, gpu=gpu)
        logger.info("EasyOCR ready.")

    def extract(self, image_path: str,
                declared_biz_name: Optional[str] = None) -> OCRResult:
        """
        Extract all visible text from image.

        Args:
            image_path:        Path to captured image
            declared_biz_name: Business name from loan application (for fuzzy matching)

        Returns:
            OCRResult
        """
        result = OCRResult()

        try:
            # EasyOCR returns [(bbox, text, confidence), ...]
            raw = self.reader.readtext(image_path, detail=1, paragraph=False)

            for (bbox, text, confidence) in raw:
                if confidence >= 0.4 and len(text.strip()) >= 2:
                    result.extracted_texts.append(text.strip())

            result.full_text = " | ".join(result.extracted_texts)

            # Fuzzy match against declared business name
            if declared_biz_name and result.extracted_texts:
                result.business_name_match_score, result.matched_name = \
                    self._fuzzy_match(declared_biz_name, result.extracted_texts)

        except Exception as e:
            logger.error(f"EasyOCR error: {e} — trying Tesseract fallback")
            result = self._tesseract_fallback(image_path, declared_biz_name)

        logger.info(f"OCR: {len(result.extracted_texts)} regions. "
                    f"Match score: {result.business_name_match_score:.2f}")
        return result

    @staticmethod
    def _tesseract_fallback(image_path: str,
                             declared_biz_name: Optional[str]) -> OCRResult:
        """Tesseract fallback when EasyOCR fails."""
        result = OCRResult()
        try:
            img = cv2.imread(image_path)
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
            text = pytesseract.image_to_string(gray, lang='eng')
            result.extracted_texts = [t.strip() for t in text.split('\n') if len(t.strip()) > 2]
            result.full_text = " | ".join(result.extracted_texts)
            if declared_biz_name and result.extracted_texts:
                result.business_name_match_score, result.matched_name = \
                    OCRModule._fuzzy_match(declared_biz_name, result.extracted_texts)
        except Exception as e:
            logger.error(f"Tesseract fallback also failed: {e}")
        return result

    @staticmethod
    def _fuzzy_match(target: str, candidates: List[str]) -> Tuple[float, Optional[str]]:
        """
        Find best fuzzy match for `target` among OCR-extracted `candidates`.
        Uses SequenceMatcher (difflib) — same as Scikit-learn text similarity logic.

        Returns:
            (best_score 0.0–1.0, best_matching_candidate)
        """
        best_score, best_match = 0.0, None
        for c in candidates:
            score = difflib.SequenceMatcher(None, target.lower(), c.lower()).ratio()
            if score > best_score:
                best_score, best_match = score, c
        return round(best_score, 3), best_match

    @staticmethod
    def preprocess_for_ocr(image_path: str) -> np.ndarray:
        """
        Image preprocessing to improve OCR accuracy on Indian signboards:
        - Upscale if width < 800px
        - CLAHE contrast enhancement (helps with varying outdoor lighting)
        - Denoise
        """
        img  = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        h, w = img.shape

        # Upscale small images
        if w < 800:
            scale = 800 / w
            img = cv2.resize(img, None, fx=scale, fy=scale,
                             interpolation=cv2.INTER_CUBIC)

        # CLAHE — improves contrast for sign text in varied lighting
        clahe    = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
        enhanced = clahe.apply(img)

        # Denoise
        denoised = cv2.fastNlMeansDenoising(enhanced, h=10)
        return denoised


# ── Demo: fuzzy matching logic ─────────────────────────────────
print("✅ OCRModule (EasyOCR + Tesseract fallback) defined.")
print("\nFuzzy Match Demo — declared name vs OCR-extracted texts:\n")

test_cases = [
    ("Patil General Store", ["PATIL GENERAL STORE", "Open 9-9", "Fresh Fruits"]),
    ("Sharma Medical",      ["Sharma Medical & Surgical", "Dr R Sharma", "Closed Sunday"]),
    ("ABC Traders",         ["XYZ Trading Co.", "Sale 50% Off", "Welcome"]),
]

for declared, extracted in test_cases:
    score, match = OCRModule._fuzzy_match(declared, extracted)
    flag = "✅ MATCH" if score >= cfg.OCR_MATCH_THRESHOLD else "⚠️  MISMATCH"
    print(f"  Declared  : '{declared}'")
    print(f"  Best OCR  : '{match}' → Score: {score:.3f}  {flag}")
    print()
```

```python
# ============================================================
# MODULE 3: SCENE CLASSIFICATION — ResNet/MobileNet (PyTorch)
# ============================================================

from torchvision import models, transforms

class SceneClassificationModule:
    """
    Classifies the scene in captured images as:
      shop / retail · residence / home · office / workplace · unclear

    Architecture: MobileNetV3-Large (fine-tuned) for speed on mobile-edge environments.
    Fallback: ResNet-50 for higher accuracy in server inference.
    (As specified in blackbook Chapter 4A: 'ResNet or MobileNet')
    """

    # Custom label mapping from ImageNet → FI-relevant scene types
    # In production: fine-tune on a labelled FI dataset (shop/residence/office images)
    SCENE_LABEL_MAP = {
        "shop": [
            "grocery store", "bakery", "butcher shop", "tobacco shop",
            "bookshop", "stationery", "hardware store"
        ],
        "residence": [
            "bedroom", "kitchen", "living room", "dining room",
            "bathroom", "balcony"
        ],
        "office": [
            "computer room", "conference room", "office",
            "lab", "library"
        ]
    }

    # ImageNet transform pipeline
    TRANSFORM = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )
    ])

    def __init__(self, model_type: str = "mobilenet"):
        """
        Args:
            model_type: 'mobilenet' (faster, mobile-suitable) or
                        'resnet' (higher accuracy, server inference)
        """
        self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

        if model_type == "mobilenet":
            self.model = models.mobilenet_v3_large(pretrained=True)
        else:
            self.model = models.resnet50(pretrained=True)

        self.model.eval()
        self.model.to(self.device)
        logger.info(f"Scene classifier ({model_type}) loaded on {self.device}")

        # Load ImageNet class labels
        self._load_imagenet_labels()

    def _load_imagenet_labels(self):
        """Load ImageNet class index → label mapping."""
        try:
            url = "https://raw.githubusercontent.com/anishathalye/imagenet-simple-labels/master/imagenet-simple-labels.json"
            resp = http_requests.get(url, timeout=5)
            self.imagenet_labels = resp.json()
        except Exception:
            logger.warning("Could not load ImageNet labels — using index numbers")
            self.imagenet_labels = [str(i) for i in range(1000)]

    def classify(self, image_path: str,
                 declared: PremisesType) -> SceneClassificationResult:
        """
        Classify scene type from captured image.

        Args:
            image_path: Path to image
            declared:   Borrower's declared premises type from application

        Returns:
            SceneClassificationResult
        """
        result = SceneClassificationResult()

        try:
            img    = Image.open(image_path).convert('RGB')
            tensor = self.TRANSFORM(img).unsqueeze(0).to(self.device)

            with torch.no_grad():
                outputs = self.model(tensor)
                probs   = torch.nn.functional.softmax(outputs, dim=1)[0]

            # Get top 5 predictions
            top5_vals, top5_idxs = torch.topk(probs, 5)
            top5 = [
                (self.imagenet_labels[idx.item()], val.item())
                for idx, val in zip(top5_idxs, top5_vals)
            ]
            result.top3_predictions = top5[:3]

            # Map top prediction to FI scene category
            top_label = top5[0][0].lower()
            result.scene_type  = self._map_to_fi_scene(top_label, top5)
            result.confidence  = round(top5[0][1], 3)
            result.matches_declared = (result.scene_type == declared)

        except Exception as e:
            logger.error(f"Scene classification error: {e}")

        logger.info(f"Scene: {result.scene_type.value} ({result.confidence:.1%}) — "
                    f"Match declared: {result.matches_declared}")
        return result

    @staticmethod
    def _map_to_fi_scene(top_label: str,
                          top5: List[Tuple[str, float]]) -> PremisesType:
        """
        Map ImageNet predictions to FI-relevant scene categories.
        In production: replace with fine-tuned model output.
        """
        all_labels = " ".join([lbl.lower() for lbl, _ in top5])

        shop_keywords = ["store", "shop", "market", "bakery", "pharmacy", "stall", "counter"]
        home_keywords = ["bedroom", "kitchen", "living", "home", "house", "couch", "bed", "dining"]
        office_keywords = ["office", "desk", "computer", "conference", "lab", "workspace"]

        if any(kw in all_labels for kw in shop_keywords):
            return PremisesType.SHOP
        if any(kw in all_labels for kw in home_keywords):
            return PremisesType.RESIDENCE
        if any(kw in all_labels for kw in office_keywords):
            return PremisesType.OFFICE
        return PremisesType.UNCLEAR


print("✅ SceneClassificationModule (ResNet/MobileNet) defined.")
print("\nUsage:")
print("  scene_mod = SceneClassificationModule('mobilenet')")
print("  result    = scene_mod.classify('shop_image.jpg', PremisesType.SHOP)")
print("  print(result.scene_type, result.confidence, result.matches_declared)")
```

```python
# ============================================================
# MODULE 5: GPS LOCATION VALIDATION
# ============================================================

class GPSValidationModule:
    """
    Validates that the GPS coordinates captured during the FI session
    correspond to the borrower's declared address.

    Flow (from DFD / System Architecture):
      Client locks GPS → sends lat/lon with each image
      → Server reverse-geocodes via Google Maps API
      → Computes distance from declared address
      → Flags if distance > GPS_MISMATCH_METRES (200m)

    From Discussion (Ch. 5B):
      93.5% GPS success rate; 6.5% failure in older reinforced concrete buildings
      → Wi-Fi positioning as fallback recommended.
    """

    GEOCODE_URL = "https://maps.googleapis.com/maps/api/geocode/json"

    def __init__(self):
        self.api_key   = cfg.GOOGLE_MAPS_API_KEY
        self.threshold = cfg.GPS_MISMATCH_METRES

    def reverse_geocode(self, lat: float, lon: float) -> str:
        """Convert GPS coordinates to human-readable address string."""
        params = {"latlng": f"{lat},{lon}", "key": self.api_key}
        try:
            resp = http_requests.get(self.GEOCODE_URL, params=params, timeout=5)
            data = resp.json()
            if data.get("status") == "OK" and data.get("results"):
                return data["results"][0]["formatted_address"]
        except Exception as e:
            logger.error(f"Reverse geocode failed: {e}")
        return "Address lookup failed"

    def geocode_address(self, address: str) -> Optional[Tuple[float, float]]:
        """Convert declared address string to (lat, lon) coordinates."""
        params = {"address": address, "key": self.api_key}
        try:
            resp = http_requests.get(self.GEOCODE_URL, params=params, timeout=5)
            data = resp.json()
            if data.get("status") == "OK" and data.get("results"):
                loc = data["results"][0]["geometry"]["location"]
                return (loc["lat"], loc["lng"])
        except Exception as e:
            logger.error(f"Geocoding failed: {e}")
        return None

    def validate(self, captured_gps: GPSCoordinates,
                 declared_address: str) -> GPSVerificationResult:
        """
        Full GPS validation:
          1. Reverse-geocode captured coordinates → human-readable address
          2. Geocode declared address → coordinates
          3. Compute geodesic distance
          4. Flag as spoof if distance > threshold (200m)

        Args:
            captured_gps:    GPS from borrower's phone during session
            declared_address: Address from loan application form

        Returns:
            GPSVerificationResult
        """
        result = GPSVerificationResult()

        # Step 1: Reverse geocode captured location
        result.captured_address = self.reverse_geocode(
            captured_gps.latitude, captured_gps.longitude
        )

        # Step 2: Geocode declared address
        declared_coords = self.geocode_address(declared_address)
        if declared_coords is None:
            logger.warning("Could not geocode declared address — GPS check skipped")
            return result

        # Step 3: Compute distance
        captured_coords = (captured_gps.latitude, captured_gps.longitude)
        result.distance_from_declared_m = round(
            geodesic(captured_coords, declared_coords).meters, 1
        )

        # Step 4: Flag decision
        result.location_match  = result.distance_from_declared_m <= self.threshold
        result.spoof_suspected = not result.location_match

        flag = "✅ OK" if result.location_match else "🚨 SPOOF SUSPECTED"
        logger.info(f"GPS: {result.distance_from_declared_m:.0f}m from declared — {flag}")
        return result


# ── Demo: GPS distance calculation (no API key needed) ─────────
print("✅ GPSValidationModule defined.")
print("\nGPS Distance Demo (geodesic distance):")
print("-" * 65)

test_scenarios = [
    (17.6806, 74.0183, 17.6810, 74.0185, "Same building (~25m)"),
    (17.6806, 74.0183, 17.6850, 74.0250, "Different locality (~650m)"),
    (17.6806, 74.0183, 18.5204, 73.8567, "Different city — Pune (~93km)"),
]

for c_lat, c_lon, d_lat, d_lon, desc in test_scenarios:
    dist = geodesic((c_lat, c_lon), (d_lat, d_lon)).meters
    flag = "✅ OK" if dist <= cfg.GPS_MISMATCH_METRES else "🚨 SPOOF FLAG"
    print(f"  {desc}")
    print(f"  Distance: {dist:,.0f}m   {flag}")
    print()
```

```python
# ============================================================
# FRAUD SIGNAL LAYER
# Rules-based scoring engine (Python + Scikit-learn logic)
# ============================================================

class FraudSignalEngine:
    """
    Aggregates outputs from all 4 AI modules + GPS into a fraud risk score.

    Score weights:
      GPS mismatch   → +30  (location is hardest to fake)
      Identity miss  → +25
      Scene mismatch → +20
      OCR mismatch   → +15
      Object absence → +10
      ─────────────────
      Total max      = 100

    Risk levels:
      0–24  → LOW    : Auto-approve eligible
      25–49 → MEDIUM : Additional verification recommended
      50+   → HIGH   : Manual review required — do not disburse
    """

    def evaluate(
        self,
        gps_result:   GPSVerificationResult,
        face_result:  FaceVerificationResult,
        scene_result: SceneClassificationResult,
        ocr_result:   OCRResult,
        yolo_signals: Dict[str, Any]
    ) -> FraudSignals:
        """
        Run all fraud checks and return aggregated FraudSignals.

        Returns:
            FraudSignals with score, risk level, and list of flag reasons
        """
        signals = FraudSignals()

        # ── FLAG 1: GPS / Location ─────────────────────────────
        # From presentation: "Location metadata or visual context
        # does not match declared address or business type"
        if gps_result.spoof_suspected:
            signals.location_flag  = True
            signals.fraud_score   += cfg.FRAUD_WEIGHT_GPS
            signals.flag_reasons.append(
                f"📍 GPS Mismatch: session location is "
                f"{gps_result.distance_from_declared_m:.0f}m from declared address "
                f"(threshold: {cfg.GPS_MISMATCH_METRES}m)"
            )

        # ── FLAG 2: Face / Identity ────────────────────────────
        # From presentation: "Face visible in captured image does
        # not match borrower's reference identity"
        if face_result.face_found and not face_result.identity_match:
            signals.identity_flag  = True
            signals.fraud_score   += cfg.FRAUD_WEIGHT_FACE
            signals.flag_reasons.append(
                f"👤 Identity Mismatch: face distance {face_result.distance:.4f} "
                f"(threshold: {cfg.FACE_SIMILARITY_THRESHOLD})"
            )
        elif not face_result.face_found:
            # No face detected — partial penalty
            signals.fraud_score  += cfg.FRAUD_WEIGHT_FACE // 2
            signals.flag_reasons.append("👤 No face detected in captured image")

        # ── FLAG 3: Scene / Environment ────────────────────────
        # From presentation: "Environment does not match declared loan purpose"
        if not scene_result.matches_declared:
            signals.scene_flag    = True
            signals.fraud_score  += cfg.FRAUD_WEIGHT_SCENE
            signals.flag_reasons.append(
                f"🏪 Scene Mismatch: AI classified '{scene_result.scene_type.value}' "
                f"but borrower declared different premises type"
            )

        # ── FLAG 4: OCR Business Name ─────────────────────────
        if ocr_result.business_name_match_score < cfg.OCR_MATCH_THRESHOLD:
            signals.ocr_flag      = True
            signals.fraud_score  += cfg.FRAUD_WEIGHT_OCR
            signals.flag_reasons.append(
                f"📝 Signage Mismatch: declared business name match score "
                f"{ocr_result.business_name_match_score:.0%} "
                f"(threshold: {cfg.OCR_MATCH_THRESHOLD:.0%})"
            )

        # ── FLAG 5: YOLO Object Absence ───────────────────────
        if yolo_signals.get("objects_flag"):
            signals.objects_flag  = True
            signals.fraud_score  += cfg.FRAUD_WEIGHT_OBJECTS
            signals.flag_reasons.append(
                f"📦 Object Flag: {yolo_signals.get('flag_reason', 'Unexpected object pattern')}"
            )

        # ── Risk Level Classification ─────────────────────────
        signals.fraud_score = min(signals.fraud_score, 100)   # cap at 100

        if signals.fraud_score >= cfg.FRAUD_HIGH_THRESHOLD:
            signals.risk_level = FraudRisk.HIGH
        elif signals.fraud_score >= cfg.FRAUD_MEDIUM_THRESHOLD:
            signals.risk_level = FraudRisk.MEDIUM
        else:
            signals.risk_level = FraudRisk.LOW

        logger.info(f"Fraud: score={signals.fraud_score}, risk={signals.risk_level.value}")
        return signals


# ── Scenario Testing ───────────────────────────────────────────
engine = FraudSignalEngine()

test_scenarios = [
    {
        "name": "✅ Clean Session — all signals clear",
        "gps":   GPSVerificationResult(spoof_suspected=False, location_match=True, distance_from_declared_m=35),
        "face":  FaceVerificationResult(face_found=True, identity_match=True, distance=0.38),
        "scene": SceneClassificationResult(scene_type=PremisesType.SHOP, matches_declared=True),
        "ocr":   OCRResult(business_name_match_score=0.87, matched_name="PATIL GENERAL STORE"),
        "yolo":  {"objects_flag": False}
    },
    {
        "name": "🚨 High Fraud — GPS + Face + Scene all flagged",
        "gps":   GPSVerificationResult(spoof_suspected=True, location_match=False, distance_from_declared_m=850),
        "face":  FaceVerificationResult(face_found=True, identity_match=False, distance=0.72),
        "scene": SceneClassificationResult(scene_type=PremisesType.RESIDENCE, matches_declared=False),
        "ocr":   OCRResult(business_name_match_score=0.11),
        "yolo":  {"objects_flag": True, "flag_reason": "Declared SHOP but no shop objects detected"}
    },
    {
        "name": "⚠️  Medium Risk — GPS only",
        "gps":   GPSVerificationResult(spoof_suspected=True, location_match=False, distance_from_declared_m=320),
        "face":  FaceVerificationResult(face_found=True, identity_match=True, distance=0.41),
        "scene": SceneClassificationResult(scene_type=PremisesType.SHOP, matches_declared=True),
        "ocr":   OCRResult(business_name_match_score=0.81, matched_name="Patil Store"),
        "yolo":  {"objects_flag": False}
    },
]

print("=" * 65)
print("FRAUD SIGNAL ENGINE — SCENARIO EVALUATION")
print("=" * 65)

for sc in test_scenarios:
    result = engine.evaluate(sc["gps"], sc["face"], sc["scene"], sc["ocr"], sc["yolo"])
    print(f"\n{sc['name']}")
    print(f"  Score: {result.fraud_score}/100  →  Risk: {result.risk_level.value}")
    for reason in result.flag_reasons:
        print(f"  • {reason}")
    if not result.flag_reasons:
        print("  • No flags raised")
```

```python
# ============================================================
# FI REPORT GENERATOR
# ============================================================

class FIReportGenerator:
    """
    Compiles all AI module outputs into a structured Digital FI Report.
    Output formats:
      - JSON  → for LOS API integration
      - Text  → for credit officer UI review
    """

    def generate(
        self,
        session:       SessionData,
        yolo_results:  List[ObjectDetectionResult],
        ocr_results:   List[OCRResult],
        scene_results: List[SceneClassificationResult],
        face_result:   FaceVerificationResult,
        gps_result:    GPSVerificationResult,
        fraud_signals: FraudSignals
    ) -> Dict[str, Any]:
        """
        Build the complete Digital FI Report dictionary.

        Returns:
            dict — fully structured report ready for JSON and LOS submission
        """
        # Aggregate objects across all images
        all_objects: Dict[str, int] = {}
        for r in yolo_results:
            for obj, count in r.object_counts.items():
                all_objects[obj] = all_objects.get(obj, 0) + count

        # Aggregate OCR texts and best match score
        all_texts:      List[str] = []
        best_ocr_score: float     = 0.0
        for r in ocr_results:
            all_texts.extend(r.extracted_texts)
            best_ocr_score = max(best_ocr_score, r.business_name_match_score)

        # Best scene result (highest confidence)
        best_scene = max(scene_results, key=lambda s: s.confidence) \
                     if scene_results else SceneClassificationResult()

        report = {
            "report_id":    str(uuid.uuid4()),
            "generated_at": datetime.datetime.utcnow().isoformat() + "Z",
            "session_id":   session.session_id,
            "report_version": "1.0",

            "applicant": {
                "name":             session.application.applicant_name,
                "loan_type":        session.application.loan_type.value,
                "declared_address": session.application.declared_address,
                "declared_premises":session.application.declared_premises.value,
                "declared_biz_name":session.application.declared_biz_name,
            },

            "session": {
                "total_images":  len(session.images),
                "started_at":    session.created_at.isoformat() + "Z",
                "gps_points":   [
                    {"lat": img.gps.latitude,
                     "lon": img.gps.longitude,
                     "ts":  img.timestamp.isoformat()}
                    for img in session.images
                ]
            },

            "ai_module_outputs": {
                "object_detection": {
                    "objects_found":   all_objects,
                    "total_count":     sum(all_objects.values()),
                    "person_present":  any(r.has_person for r in yolo_results),
                },
                "ocr": {
                    "all_extracted_texts":       all_texts,
                    "best_biz_name_match_score":  round(best_ocr_score, 3),
                },
                "scene_classification": {
                    "classified_as":    best_scene.scene_type.value,
                    "confidence":       round(best_scene.confidence, 3),
                    "matches_declared": best_scene.matches_declared,
                },
                "face_verification": {
                    "face_found":      face_result.face_found,
                    "identity_match":  face_result.identity_match,
                    "distance":        face_result.distance,
                    "face_count":      face_result.face_count,
                },
                "gps_validation": {
                    "captured_address":         gps_result.captured_address,
                    "distance_from_declared_m": gps_result.distance_from_declared_m,
                    "location_match":            gps_result.location_match,
                }
            },

            "fraud_assessment": {
                "fraud_score":  fraud_signals.fraud_score,
                "risk_level":   fraud_signals.risk_level.value,
                "flags": {
                    "location":  fraud_signals.location_flag,
                    "identity":  fraud_signals.identity_flag,
                    "scene":     fraud_signals.scene_flag,
                    "ocr":       fraud_signals.ocr_flag,
                    "objects":   fraud_signals.objects_flag,
                },
                "flag_reasons": fraud_signals.flag_reasons,
            },

            "recommendation": self._recommendation(fraud_signals.risk_level)
        }

        return report

    @staticmethod
    def _recommendation(risk: FraudRisk) -> str:
        return {
            FraudRisk.LOW:    "AUTO-APPROVE ELIGIBLE: All signals consistent. Proceed to credit assessment.",
            FraudRisk.MEDIUM: "ADDITIONAL VERIFICATION RECOMMENDED: One or more signals require review.",
            FraudRisk.HIGH:   "MANUAL REVIEW REQUIRED: Multiple fraud signals. Do not disburse without in-person investigation."
        }[risk]

    @staticmethod
    def format_text(report: Dict[str, Any]) -> str:
        """Human-readable FI report for credit officer UI."""
        ai = report["ai_module_outputs"]
        fa = report["fraud_assessment"]

        lines = [
            "═" * 62,
            "        DIGITAL FIELD INVESTIGATION REPORT",
            "═" * 62,
            f"Report ID    : {report['report_id'][:8]}...",
            f"Generated    : {report['generated_at']}",
            f"Session      : {report['session_id'][:8]}...",
            "",
            "── APPLICANT ────────────────────────────────────────────",
            f"Name         : {report['applicant']['name']}",
            f"Loan Type    : {report['applicant']['loan_type'].upper()}",
            f"Address      : {report['applicant']['declared_address']}",
            f"Premises     : {report['applicant']['declared_premises'].upper()}",
            f"Business     : {report['applicant']['declared_biz_name'] or 'N/A'}",
            "",
            "── AI MODULE RESULTS ────────────────────────────────────",
            f"Objects      : {ai['object_detection']['total_count']} detections  "
            f"Person present: {'YES' if ai['object_detection']['person_present'] else 'NO'}",
            f"OCR Match    : {ai['ocr']['best_biz_name_match_score']:.0%}",
            f"Scene        : {ai['scene_classification']['classified_as'].upper()} "
            f"({ai['scene_classification']['confidence']:.0%}) — "
            f"{'✅ MATCH' if ai['scene_classification']['matches_declared'] else '⚠️  MISMATCH'}",
            f"Face         : {'✅ MATCH' if ai['face_verification']['identity_match'] else '⚠️  MISMATCH'} "
            f"(distance: {ai['face_verification']['distance']:.4f})",
            f"GPS          : {ai['gps_validation']['distance_from_declared_m']:.0f}m from declared — "
            f"{'✅ OK' if ai['gps_validation']['location_match'] else '🚨 MISMATCH'}",
            "",
            "── FRAUD ASSESSMENT ─────────────────────────────────────",
            f"Score        : {fa['fraud_score']}/100",
            f"Risk Level   : {fa['risk_level']}",
        ]

        if fa["flag_reasons"]:
            lines.append("Flags        :")
            for r in fa["flag_reasons"]:
                lines.append(f"  • {r}")
        else:
            lines.append("Flags        : None raised")

        lines += [
            "",
            "── RECOMMENDATION ───────────────────────────────────────",
            report["recommendation"],
            "═" * 62
        ]

        return "\n".join(lines)


# ── Demo: generate mock report ─────────────────────────────────
gen = FIReportGenerator()

mock_session = SessionData(
    application=sample_app,
    images=[CapturedImage(
        image_index=1, prompt_shown="Capture shop exterior",
        s3_key="sessions/demo/img001.jpg",
        timestamp=datetime.datetime.utcnow(),
        gps=GPSCoordinates(latitude=17.6806, longitude=74.0183)
    )]
)

mock_report = gen.generate(
    session       = mock_session,
    yolo_results  = [ObjectDetectionResult(object_counts={'person':1,'bottle':3,'cup':2}, has_person=True)],
    ocr_results   = [OCRResult(extracted_texts=['PATIL GENERAL STORE','Open 9-9'], business_name_match_score=0.87)],
    scene_results = [SceneClassificationResult(scene_type=PremisesType.SHOP, confidence=0.91, matches_declared=True)],
    face_result   = FaceVerificationResult(face_found=True, identity_match=True, distance=0.38),
    gps_result    = GPSVerificationResult(captured_address='MG Road, Satara', distance_from_declared_m=35, location_match=True),
    fraud_signals = FraudSignals(fraud_score=0, risk_level=FraudRisk.LOW)
)

print(gen.format_text(mock_report))
```
